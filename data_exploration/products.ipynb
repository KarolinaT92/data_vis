```python
from read_data import cleaned_df
import plotly.express as px
import plotly.graph_objects as go
import numpy as np
import pandas as pd
```

```python
# Profit margin
cleaned_df["Profit Margin (%)"] = (cleaned_df["Profit"] / cleaned_df["Sales"]) * 100
cleaned_df["Profit Margin (%)"] = cleaned_df["Profit Margin (%)"].round(2)
cleaned_df[["Sales", "Profit"]] = cleaned_df[["Sales", "Profit"]].round(2)
cleaned_df['Month'] = cleaned_df['Order Date'].dt.month
cleaned_df['Year'] = cleaned_df['Order Date'].dt.year
```

```python
df = cleaned_df.copy()

df["Order Date"] = pd.to_datetime(df["Order Date"])

# Filter for year 2017
sales_2017 = df[df["Order Date"].dt.year == 2017]["Sales"].sum()
sales_2017
```

```python
# Filter for year 2017
profit_2017 = df[df["Order Date"].dt.year == 2017]["Profit"].sum()

print(f"Total profit in 2017: ${profit_2017:,.2f}")
```

```python
# Filter for year 2017
df_2017 = df[df["Year"] == 2017]

# Group by Category
grouped = (
    df_2017.groupby("Category", as_index=False)
    .agg({
        "Sales": "sum",
        "Profit": "sum",
        "Quantity": "sum"
    })
)

# Bubble chart
fig = px.scatter(
    grouped,
    x="Sales",
    y="Profit",
    size="Quantity",
    color="Category",
    hover_name="Category",
    text="Category",  # ✅ show category labels on bubbles
    size_max=60,
    title="Category Value Distribution (Sales, Profit, Quantity) in 2017",
    labels={
        "Sales": "Total Sales",
        "Profit": "Total Profit",
        "Quantity": "Total Quantity"
    }
)

# Make text appear centered on bubbles
fig.update_traces(
    textposition="middle center",
    textfont=dict(size=12, color="black"),
    opacity=0.85
)

# ✅ Only show gridlines where the bubbles exist
sales_ticks = sorted(grouped["Sales"].round(0).unique())
profit_ticks = sorted(grouped["Profit"].round(0).unique())

fig.update_layout(
    showlegend=False,
    xaxis=dict(
        title="Sales",
        showgrid=True,
        tickvals=sales_ticks,  # ✅ gridlines at bubble X positions
        gridcolor="lightgrey",
        gridwidth=0.5,
        griddash="dot"
    ),
    yaxis=dict(
        title="Profit",
        showgrid=True,
        tickvals=profit_ticks,  # ✅ gridlines at bubble Y positions
        gridcolor="lightgrey",
        gridwidth=0.5,
        griddash="dot"
    ),
    plot_bgcolor="white"
)

fig.show()
```

```python
# Aggregate by product (keep category info for hover)
grouped_products = (
    df_2017.groupby(["Product Name", "Category", "Sub-Category"], as_index=False)
    .agg({"Sales": "sum", "Profit": "sum"})
)

# Top 10 by Profit
top10 = grouped.sort_values("Profit", ascending=False).head(10)

# Mid-range blue scale (avoid white/black extremes)
custom_blue_scale = [
    [0.0, "#99c9ff"],
    [0.5, "#4da6ff"],
    [1.0, "#0059b3"]
]
# Sort by Profit descending and take Top 10
# Top 10 most profitable
top10_products = grouped_products.sort_values("Profit", ascending=False).head(10)

# 🎨 Custom mid-range blue color scale (avoids pure white or black)
custom_blue_scale = [
    [0.0, "#99c9ff"],  # light-mid blue
    [0.5, "#4da6ff"],  # medium blue
    [1.0, "#0059b3"]  # deep strong blue (not navy)
]

# Horizontal bar chart
fig = px.bar(
    top10_products,
    x="Profit",
    y="Product Name",
    orientation="h",
    text="Profit",
    title="Top 10 Most Profitable Products (2017)",
    labels={"Profit": "Total Profit", "Product Name": "Product Name"},
    color="Profit",
    color_continuous_scale=custom_blue_scale
)

# Compute headroom for labels so they don't clip
x_max = top10_products["Profit"].max()

# Display profit values clearly and fully visible
fig.update_traces(
    texttemplate="%{text:,.0f}",
    textposition="outside",
    textfont=dict(size=13, color="black"),
    cliponaxis=False,
    hovertemplate=(
        "<b>%{y}</b><br>"
        "Profit: %{x:,.2f}<br>"
        "Category: %{customdata[0]}<br>"
        "Sub-Category: %{customdata[1]}<br>"
        "Sales: %{customdata[2]:,.2f}"
    ),
    customdata=top10_products[["Category", "Sub-Category", "Sales"]],
)

# Layout styling
fig.update_layout(
    showlegend=False,
    coloraxis_showscale=False,  # hide color bar
    yaxis=dict(
        categoryorder="total ascending",
        showgrid=False
    ),
    xaxis=dict(
        title="Total Profit",
        showgrid=True,
        gridcolor="lightgrey",
        gridwidth=0.4,
        automargin=True,
        range=[0, x_max * 1.18],  # ✅ headroom so full label stays visible
    ),
    margin=dict(l=80, r=140, t=60, b=40),
    plot_bgcolor="white"
)

fig.show()
```

```python

df_20172 = df[df["Year"] == 2017]
grouped2 = (
    df_20172.groupby(["Product Name", "Category", "Sub-Category"], as_index=False)
    .agg({"Sales": "sum", "Profit": "sum"})
)

# Top 10 by Profit
top102 = grouped2.sort_values("Profit", ascending=False).head(10)

# Mid-range blue scale (avoid white/black extremes)
custom_blue_scale = [
    [0.0, "#99c9ff"],
    [0.5, "#4da6ff"],
    [1.0, "#0059b3"]
]

# Colors
blue_title = "#0059b3"
orange = "orange"
orange_dark = "#b34700"  # label color

# Profit bars (full-width)
profit_trace = go.Bar(
    x=top102["Profit"],
    y=top102["Product Name"],
    orientation="h",
    name="Profit",
    marker=dict(color=top102["Profit"], opacity=0.8, colorscale=custom_blue_scale, showscale=False),
    width=0.8,  # thicker outer bar
    text=top102["Profit"].map("{:,.0f}".format),
    textposition="none",  #"outside",
    textfont=dict(size=13, color="#003366"),
    texttemplate=" %{text}",
    cliponaxis=False,
    customdata=top102[["Category", "Sub-Category", "Sales"]],
    hovertemplate=(
        "<b>%{y}</b><br>"
        "Profit: %{x:,.2f}<br>"
        "Category: %{customdata[0]}<br>"
        "Sub-Category: %{customdata[1]}<br>"
        "Sales: %{customdata[2]:,.2f}"
    ),
    # opacity=0.95
)

# Sales mini-bars (narrower, overlaid)
sales_trace = go.Bar(
    x=top102["Sales"],
    y=top102["Product Name"],
    text=top102["Sales"].map("{:,.0f}".format),  
    textposition="outside",
    outsidetextfont=dict(size=10, color=orange_dark, family="Arial Black"),
    # textfont=dict(size=26, color=orange_dark, family="Arial Black"),
    insidetextanchor="start", 
    cliponaxis=False,  
    orientation="h",
    name="Sales",
    marker=dict(color=orange),
    width=0.2,  # thinner inner bar so it sits inside Profit
    hovertemplate="Sales: %{x:,.2f}<extra></extra>",
    opacity=0.7,
    xaxis="x2",
)

fig = go.Figure([profit_trace, sales_trace])

# Headroom so profit labels don't clip
x_max = top102["Profit"].max()
x_max_sales = top102["Sales"].max()

fig.update_layout(
    title="Top 10 Most Profitable Products and their Sales (2017) ", #— Profit (outer) & Sales (inner)
    barmode="overlay",  # ✅ overlay bars so red sits inside blue
    uniformtext=dict(mode="show", minsize=4),  
    yaxis=dict(categoryorder="total ascending", showgrid=False, title=""),
    xaxis=dict(
        title="Total Profit ($)",
        color=blue_title,
        tickfont=dict(color=blue_title),
        showgrid=True, gridcolor="lightgrey", gridwidth=0.4,
        range=[0, x_max * 1.35]  # headroom for outside labels
    ),
    xaxis2=dict(
        title="Total Sales ($)",
        tickfont=dict(color=orange),
        color=orange,
        overlaying="x",
        side="top",
        showgrid=False,
        range=[0, x_max_sales * 1.18]  # ✅ separate scale for Sales
    ),
    showlegend=False,  # labels + colors are self-explanatory
    margin=dict(l=110, r=180, t=140, b=40),
    plot_bgcolor="white",
    hovermode="y unified"
)

fig.show()
```
